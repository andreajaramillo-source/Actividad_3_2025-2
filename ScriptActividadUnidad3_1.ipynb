```
# Definición de librería para crear gráficos en python

import matplotlib.pyplot as plt
```

```
# Datos de las l¿listas usos y áreas

usos = ['Cultivo de secano', 'Cultivo irrigado', 'Urbano', 'Suburbano', 'Pastizales', 'Agua', 'Humedales', 'Bosques', 'Arbustos']
areas = [15, 10, 10, 2, 8, 12, 3, 25, 15]
```

#Funciones de Listas

### Se requiere que a continuación construya el script que le permita obtener e imprimir usando las funciones de las listas:

*   La menor área
*   La mayor área
*   El área total


```
# Funciones de listas

def encontrar_maximo(areas):
  return max(areas)

def encontrar_minimo(areas):
  return min(areas)

def calcular_suma(areas):
  return sum(areas)
```

```
# Funciones de lista area_maxima, area_mnimia, suma_total_area

mayor_area = max(areas)
print("El dato mayor en la lista de área es:", mayor_area)

menor_area = min(areas)
print("El dato menor en la lista de área es:", menor_area)

suma_total_areas = sum(areas)
print("La suma total de las áreas es:", suma_total_areas)
```

#Gráfico Estadístico

### Construya un gráfico estadístico de torta que contenga la cuadrícula de fondo, etiquetas, escala gráfica y el título del gráfico.

```
# Creación del gráfico de barras

plt.figure(figsize=(10, 6))
plt.bar(usos, areas, color= 'skyblue')
plt.title("Distribución de Usos de Suelo", fontsize=16)
plt.xlabel("Uso de suelo", fontsize=12)
plt.ylabel("Áreas", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
